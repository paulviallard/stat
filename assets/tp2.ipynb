Jusqu’ici, on a entraîné un modèle sans se soucier de sa performance. Pour cela, nous avons d'abord besoin de mesurer la qualité du modèle. 

Considérons $y$ et $y'$ deux prédictions (par exemple, $y'$ est la classe prédite par le modèle et $y$ la classe réelle). Nous pouvons utiliser la fonction de perte $0-1$ pour évaluer la qualité d'une prédiction :
$$

**3.a.** Calculer le risque sur les données d'entraînement de l'arbre de décision :

$$

Cependant, le modèle peut simplement mémoriser les données et ne pas bien fonctionner sur de nouvelles observations. Il faut donc évaluer ses performances sur un autre jeu de données, représentant de nouvelles observations.
